### D1.4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.​

```python
df["Marks_raw"] = df["Marks"].astype("string")
df["Marks"] = df["Marks_raw"].str.extract(r"(-?\d+(?:\.\d+)?)")[0]
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
```

```python
display(df[["Student_ID", "Student_Name", "Marks_raw", "Marks"]])
```

### Explonation:

### D1.5.  Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.​

```python
df["Mark_Status"] = "Available"

df.loc[df["Marks_raw"].str.lower() == "absent","Mark_Status"] = "Absent"
df.loc[df["Marks"].isna() &(df["Mark_Status"] != "Absent"),"Mark_Status"] = "Not Entered / NA"
display(df[["Student_ID","Student_Name","Marks_raw","Marks","Mark_Status"]])
```

### Explanation:

I investigate the missing marks instead of deleting the records immediately. Absent means the student was absent, while blank or NA may mean that the mark was not entered or is unavailable. Deleting missing records without investigating can bias the analysis if the missingness is related to attendance or student performance.

### D1.6. Identify marks outside the valid range of 0–100 and decide how they should be handled.​

```python
invalid_mask = (df["Marks"] < 0) | (df["Marks"] > 100)
invalid_marks = df.loc[invalid_mask].copy()

print("Invalid marks:")
display(invalid_marks[["Student_ID", "Student_Name", "Marks"]])

df.loc[invalid_mask, "Mark_Status"] = "Invalid"
df.loc[invalid_mask, "Marks"] = pd.NA
```

### Explanation:

I check whether the marks are between 0 and 100. A value outside this range is flagged as Invalid and its numeric value is set to missing until it can be investigated. I keep the original record and report how many invalid marks were identified.
